```python
missing_analysis = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_analysis = missing_analysis.sort_values(
    by="Missing_Percentage",
    ascending=False
)

missing_analysis
```

```python
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
print(
    "Duplicate percentage:",
    round(duplicate_count / len(df) * 100, 2),
    "%"
)
```

```python
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())
print("Negative unit prices:", (df["UnitPrice"] < 0).sum())
print("Zero unit prices:", (df["UnitPrice"] == 0).sum())
```

```python
print(
    "Invoices starting with C:",
    df["InvoiceNo"].astype(str).str.startswith("C").sum()
)
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

print("Earliest invoice date:", df["InvoiceDate"].min())
print("Latest invoice date:", df["InvoiceDate"].max())
print("Invalid dates:", df["InvoiceDate"].isna().sum())
```

```python
plt.figure(figsize=(8, 5))

df["Quantity"].dropna().hist(bins=50)

plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

df["UnitPrice"].dropna().hist(bins=50)

plt.title("Distribution of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Create a separate dataset for visualization only
quantity_data = df.loc[
    (df["Quantity"] > 0) &
    (df["Quantity"] <= df["Quantity"].quantile(0.99)),
    "Quantity"
]

plt.figure(figsize=(8, 5))
plt.hist(quantity_data, bins=40, edgecolor="black")
plt.title("Distribution of Quantity (Up to 99th Percentile)")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
price_data = df.loc[
    (df["UnitPrice"] > 0) &
    (df["UnitPrice"] <= df["UnitPrice"].quantile(0.99)),
    "UnitPrice"
]

plt.figure(figsize=(8, 5))
plt.hist(price_data, bins=40, edgecolor="black")
plt.title("Distribution of Unit Price (Up to 99th Percentile)")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Country-wise transaction records
country_analysis = (
    df.groupby("Country")
    .agg(
        Total_Transactions=("InvoiceNo", "nunique"),
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Quantity", lambda x: 0)
    )
    .reset_index()
)

country_analysis.head(10)
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

country_analysis = (
    df.groupby("Country")
    .agg(
        Total_Invoices=("InvoiceNo", "nunique"),
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Revenue", "sum")
    )
    .sort_values("Total_Revenue", ascending=False)
    .reset_index()
)

country_analysis.head(10)
```

```python
# Top 10 countries by revenue

top_10_countries = country_analysis.head(10)

plt.figure(figsize=(12, 6))

plt.bar(
    top_10_countries["Country"],
    top_10_countries["Total_Revenue"]
)

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

## Step 11: Analysis Strategy

| Analysis Area | Method | Purpose |
|---|---|---|
| Dataset structure | `shape`, `columns`, `info()` | Understand rows, columns, and data types |
| Summary statistics | `describe()` | Understand numeric distributions |
| Missing values | `isnull().sum()` | Identify incomplete data |
| Duplicate records | `duplicated().sum()` | Identify repeated rows |
| Anomalies | Quantity and UnitPrice checks | Investigate unusual values |
| Revenue analysis | Quantity × UnitPrice | Estimate transaction revenue |
| Country analysis | `groupby()` and aggregation | Compare revenue across countries |
| Visualisation | Histograms and bar charts | Communicate patterns clearly |
| Time analysis | Monthly aggregation and line chart | Investigate revenue trends |
| Interpretation | Evidence-based observations | Support recommendations |

### Tools
- Python
- pandas and NumPy
- Matplotlib and Seaborn
- Jupyter Notebook in Visual Studio Code

### Limitations
1. Missing CustomerID values may prevent complete customer-level analysis.
2. Negative quantities and cancelled invoices require careful interpretation.
3. Extreme values can affect summary statistics and visualisations.
4. Observed relationships do not automatically establish causation.
5. Business recommendations should be based on verified results, not assumptions.

```python
# Create a copy for monthly analysis
monthly_df = df.copy()

# Extract month from invoice date
monthly_df["Month"] = (
    monthly_df["InvoiceDate"].dt.to_period("M")
)

# Calculate monthly revenue
monthly_revenue = (
    monthly_df.groupby("Month")["Revenue"]
    .sum()
    .reset_index()
)

# Convert month to string for plotting
monthly_revenue["Month"] = (
    monthly_revenue["Month"].astype(str)
)

monthly_revenue
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_revenue["Month"],
    monthly_revenue["Revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
